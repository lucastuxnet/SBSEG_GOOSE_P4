# 🚀 Pipeline de Transformação: GOOSE IDS Rules → Código P4\n\n## De regras Python geradas por LLM para switch programável\n\nEste notebook implementa o pipeline completo para transformar as regras de detecção\ndo arquivo `rules.py` em código P4 executável em switches programáveis.\n\n### Fluxo de trabalho:\n1. Carregar e analisar as regras Python\n2. Gerar código P4 com suporte a estado\n3. Criar tráfego de teste (normal e ataques)\n4. Validar regras localmente\n5. Preparar ambiente para execução real\n\n---

````python
# Célula 1: Setup e Verificação do Ambiente
import subprocess
import sys
import os
import platform
import inspect
import re
from typing import List, Dict, Any
from pathlib import Path

def verificar_ambiente():
    """Verifica e prepara o ambiente"""
    print("🔍 Verificando ambiente...\n")
    
    # Verifica Python
    print(f"✅ Python: {sys.version}")
    
    # Verifica rules.py
    if not os.path.exists('rules.py'):
        print("❌ Arquivo rules.py não encontrado!")
        return False
    print("✅ rules.py encontrado")
    
    # Verifica/instala pacotes necessários
    pacotes = ['scapy', 'numpy', 'pandas', 'matplotlib']
    for pacote in pacotes:
        try:
            __import__(pacote)
            print(f"✅ {pacote} disponível")
        except ImportError:
            print(f"📥 Instalando {pacote}...")
            subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", pacote])
    
    print("\n✅ Ambiente verificado com sucesso!\n")
    return True

verificar_ambiente()
````

````python
# Célula 2: Carregar e Analisar Regras do rules.py
import rules

def carregar_regras() -> List[Dict[str, Any]]:
    """Extrai todas as regras do arquivo rules.py"""
    regras = []
    
    for name, func in inspect.getmembers(rules, inspect.isfunction):
        if name.startswith('rule_'):
            source = inspect.getsource(func)
            # Limpa a assinatura e docstring
            lines = source.splitlines()
            cleaned_lines = []
            for line in lines:
                stripped = line.strip()
                if stripped and not stripped.startswith(('def ', '"""', "'''")):
                    cleaned_lines.append(line)
            
            # Extrai a condição
            condition = ''
            for line in cleaned_lines:
                if 'return' in line:
                    condition = line.replace('return', '').strip()
                    break
            
            regras.append({
                'name': name,
                'source': '\n'.join(cleaned_lines),
                'function': func,
                'condition': condition
            })
    
    return regras

regras = carregar_regras()
print(f"✅ {len(regras)} regras carregadas de rules.py\n")
print("Regras encontradas:")
for i, r in enumerate(regras, 1):
    print(f"  {i:2d}. {r['name']}")
    if len(r['condition']) > 80:
        print(f"      Condição: {r['condition'][:80]}...")
    elif r['condition']:
        print(f"      Condição: {r['condition']}")
````

````python
# Célula 3: Gerador de Código P4 AUTO-CONTIDO (sem dependências externas)

def converter_condicao_p4(condicao_python: str) -> str:
    """Converte condição Python para sintaxe P4"""
    if not condicao_python:
        return "false"
    
    cond = condicao_python
    
    # Substitui packet.get("campo", default) por hdr.goose.campo
    import re
    pattern = r'packet\.get\(["\'](\w+)["\'],\s*(\d+)\)'
    cond = re.sub(pattern, r'hdr.goose.\1', cond)
    
    # Remove aspas
    cond = cond.replace('"', '').replace("'", "")
    
    # Substitui operadores lógicos
    cond = cond.replace(' and ', ' && ')
    cond = cond.replace(' or ', ' || ')
    cond = cond.replace(' and(', ' && (')
    cond = cond.replace(' or(', ' || (')
    
    # Corrige campos para minúsculas
    cond = cond.replace('SqNum', 'sqnum')
    cond = cond.replace('StNum', 'stnum')
    cond = cond.replace('tDiff', 'tdiff')
    cond = cond.replace('cbStatus', 'cbstatus')
    cond = cond.replace('stDiff', 'st_diff')
    cond = cond.replace('sqDiff', 'sq_diff')
    cond = cond.replace('timestampDiff', 'timestamp_diff')
    cond = cond.replace('timeFromLastChange', 'time_from_last_change')
    
    return cond

def gerar_p4_auto_contido(regras: List[Dict[str, Any]]) -> str:
    """Gera código P4 completamente auto-contido"""
    
    p4_code = '''#include <core.p4>
#include <v1model.p4>

/* Definicao do cabecalho Ethernet */
header ethernet_t {
    bit<48> dstAddr;
    bit<48> srcAddr;
    bit<16> etherType;
}

/* Cabecalho GOOSE */
header goose_h_t {
    bit<48> dstAddr;
    bit<48> srcAddr;
    bit<16> ethertype;
    bit<16> appid;
    bit<16> length;
    bit<16> reserved1;
    bit<16> reserved2;
    bit<32> stnum;
    bit<32> sqnum;
    bit<32> tdiff;
    bit<8>  cbstatus;
    bit<24> padding;
}

/* Metadados para estado */
struct metadata_t {
    bit<32> st_diff;
    bit<32> sq_diff;
    bit<32> timestamp_diff;
    bit<32> time_from_last_change;
}

/* Estrutura de headers */
struct headers_t {
    ethernet_t   ethernet;
    goose_h_t    goose;
}

/* Parser */
parser MyParser(packet_in packet,
                out headers_t hdr,
                inout metadata_t meta,
                inout standard_metadata_t standard_metadata) {
    
    state start {
        packet.extract(hdr.ethernet);
        transition select(hdr.ethernet.etherType) {
            16'h88B8: parse_goose;
            default: accept;
        }
    }
    
    state parse_goose {
        packet.extract(hdr.goose);
        transition accept;
    }
}

control MyVerifyChecksum(inout headers_t hdr, inout metadata_t meta) {
    apply { }
}

/* Pipeline de Ingress */
control MyIngress(inout headers_t hdr,
                  inout metadata_t meta,
                  inout standard_metadata_t standard_metadata) {
    
    // Acao para descartar pacote
    action drop_packet() {
        mark_to_drop(standard_metadata);
    }
    
    // Acao para encaminhar pacote
    action forward_packet() {
        standard_metadata.egress_spec = 2;
    }
    
    // Acoes especificas para cada tipo de ataque
'''
    
    # Adiciona ações para cada regra
    for regra in regras:
        p4_code += f"""
    action {regra['name']}_action() {{
        // Ataque detectado: {regra['name']}
        drop_packet();
    }}
"""
    
    p4_code += """
    apply {
        // Inicializa metadados
        meta.st_diff = 0;
        meta.sq_diff = 0;
        meta.timestamp_diff = 0;
        meta.time_from_last_change = 0;
        
"""
    
    # Adiciona condições para cada regra
    for regra in regras:
        if regra['condition']:
            cond_p4 = converter_condicao_p4(regra['condition'])
            # Simplifica condição para evitar erros de sintaxe
            cond_p4 = cond_p4.replace('(', ' ( ').replace(')', ' ) ')
            cond_p4 = ' '.join(cond_p4.split())
            
            # Limita o tamanho da condição
            if len(cond_p4) > 200:
                cond_p4 = "false  // Regra muito complexa"
            
            p4_code += f"""
        // Regra: {regra['name']}
        if ({cond_p4}) {{
            {regra['name']}_action();
        }}
"""
    
    p4_code += """
    }
}

control MyEgress(inout headers_t hdr,
                 inout metadata_t meta,
                 inout standard_metadata_t standard_metadata) {
    apply { }
}

control MyComputeChecksum(inout headers_t hdr, inout metadata_t meta) {
    apply { }
}

control MyDeparser(packet_out packet, in headers_t hdr) {
    apply {
        packet.emit(hdr.ethernet);
        packet.emit(hdr.goose);
    }
}

V1Switch(
    MyParser(),
    MyVerifyChecksum(),
    MyIngress(),
    MyEgress(),
    MyComputeChecksum(),
    MyDeparser()
) main;
"""
    
    return p4_code

# Gera o código P4
print("🔨 Gerando código P4 auto-contido...")
codigo_p4 = gerar_p4_auto_contido(regras)

# Salva o arquivo
with open('goose_ids.p4', 'w') as f:
    f.write(codigo_p4)

print("✅ Código P4 gerado com sucesso!")
print(f"📄 Arquivo: goose_ids.p4")
print(f"📊 Tamanho: {len(codigo_p4)} caracteres")
print(f"🎯 Regras implementadas: {len(regras)}")

# Mostra a estrutura do cabeçalho
print("\n📋 Estrutura do cabeçalho Ethernet definido:")
print("-" * 40)
print("header ethernet_t {")
print("    bit<48> dstAddr;")
print("    bit<48> srcAddr;")
print("    bit<16> etherType;")
print("}")
print("\n📋 Estrutura do cabeçalho GOOSE:")
print("-" * 40)
print("header goose_h_t {")
print("    bit<48> dstAddr;")
print("    bit<48> srcAddr;")
print("    bit<16> ethertype;")
print("    bit<16> appid;")
print("    bit<16> length;")
print("    ...")
print("    bit<32> stnum;")
print("    bit<32> sqnum;")
print("    bit<32> tdiff;")
print("    bit<8>  cbstatus;")
print("}")

# Verifica regras convertidas
print("\n📝 Exemplo de regras convertidas:")
print("=" * 60)
for i, regra in enumerate(regras[:3]):
    if regra['condition']:
        cond_p4 = converter_condicao_p4(regra['condition'])
        print(f"\n{i+1}. {regra['name']}:")
        print(f"   Python: {regra['condition'][:80]}...")
        print(f"   P4:     if ({cond_p4[:80]})...")
````

````python
# Célula 4: Gerador de Tráfego GOOSE para Teste

from scapy.all import Ether, Raw, wrpcap
import random

def criar_pacote_goose(sqnum, stnum, tdiff=1000, cbstatus=1):
    """Cria um pacote GOOSE simulado"""
    dst_mac = "01:0C:CD:01:00:00"  # Multicast GOOSE
    src_mac = "00:11:22:33:44:55"
    
    # Simula payload GOOSE (formato simplificado)
    goose_payload = (
        b'\x00\x00' +  # APPID
        b'\x00\x64' +  # Comprimento (100 bytes)
        stnum.to_bytes(4, 'big') +
        sqnum.to_bytes(4, 'big') +
        tdiff.to_bytes(4, 'big') +
        bytes([cbstatus]) +
        b'\x00' * 80  # Padding
    )
    
    return Ether(dst=dst_mac, src=src_mac, type=0x88B8) / Raw(load=goose_payload)

def gerar_trafego_teste():
    """Gera arquivos PCAP para teste"""
    
    print("🔧 Gerando tráfego de teste...\n")
    
    # Tráfego normal
    normal = []
    for i in range(100):
        sqnum = i % 256
        stnum = 1000 + (i * 10)
        normal.append(criar_pacote_goose(sqnum, stnum, tdiff=100, cbstatus=1))
    
    # Tráfego Grayhole (SqNum baixo + tDiff alto)
    grayhole = []
    for i in range(50):
        grayhole.append(criar_pacote_goose(sqnum=3, stnum=1000, tdiff=2000, cbstatus=1))
    
    # Tráfego High StNum (StNum > 2000)
    high_stnum = []
    for i in range(50):
        high_stnum.append(criar_pacote_goose(sqnum=50, stnum=2500, tdiff=100, cbstatus=1))
    
    # Tráfego Injection (SqNum > 80)
    injection = []
    for i in range(50):
        injection.append(criar_pacote_goose(sqnum=100, stnum=10, tdiff=100, cbstatus=2))
    
    # Salva arquivos
    wrpcap('traffic_normal.pcap', normal)
    wrpcap('traffic_grayhole.pcap', grayhole)
    wrpcap('traffic_high_stnum.pcap', high_stnum)
    wrpcap('traffic_injection.pcap', injection)
    
    print("✅ Tráfego de teste gerado:")
    print(f"   - traffic_normal.pcap: {len(normal)} pacotes (normal)")
    print(f"   - traffic_grayhole.pcap: {len(grayhole)} pacotes (Grayhole)")
    print(f"   - traffic_high_stnum.pcap: {len(high_stnum)} pacotes (High StNum)")
    print(f"   - traffic_injection.pcap: {len(injection)} pacotes (Injection)")
    
    return len(normal) + len(grayhole) + len(high_stnum) + len(injection)

total_pacotes = gerar_trafego_teste()
print(f"\n📊 Total de pacotes gerados: {total_pacotes}")
print("\n💡 Dica: Use 'tcpreplay' para enviar esses PCAPs para o switch P4")
````

````python
# Célula 5: Validação das Regras com PCAPs

from scapy.all import rdpcap

def extrair_campos_pacote(packet):
    """Extrai campos GOOSE de um pacote"""
    try:
        raw_data = bytes(packet[Raw].load)
        if len(raw_data) >= 17:
            return {
                'SqNum': int.from_bytes(raw_data[8:12], 'big'),
                'StNum': int.from_bytes(raw_data[4:8], 'big'),
                'tDiff': int.from_bytes(raw_data[12:16], 'big'),
                'cbStatus': raw_data[16],
                'stDiff': 0,
                'sqDiff': 0,
                'timestampDiff': 0.1,
                'timeFromLastChange': 0,
                'delay': 0.0
            }
    except:
        pass
    return None

def validar_regras():
    """Valida as regras contra os PCAPs gerados"""
    
    print("🔍 Validando regras contra PCAPs...\n")
    print("=" * 60)
    
    arquivos_pcap = ['traffic_normal.pcap', 'traffic_grayhole.pcap', 
                     'traffic_high_stnum.pcap', 'traffic_injection.pcap']
    
    for pcap_file in arquivos_pcap:
        if not os.path.exists(pcap_file):
            print(f"⚠️  {pcap_file} não encontrado. Execute a Célula 4 primeiro.")
            continue
            
        packets = rdpcap(pcap_file)
        alerts = {regra['name']: 0 for regra in regras}
        
        for pkt in packets:
            if Raw in pkt and pkt[Ether].type == 0x88B8:
                campos = extrair_campos_pacote(pkt)
                if campos:
                    for regra in regras:
                        try:
                            if regra['function'](campos):
                                alerts[regra['name']] += 1
                        except Exception as e:
                            pass
        
        total_alerts = sum(alerts.values())
        tipo = pcap_file.replace('traffic_', '').replace('.pcap', '').upper()
        
        print(f"\n📁 {tipo}:")
        print(f"   Pacotes: {len(packets)}")
        print(f"   Alertas: {total_alerts}")
        
        if total_alerts > 0:
            print("   Regras disparadas:")
            for rule, count in sorted(alerts.items(), key=lambda x: x[1], reverse=True):
                if count > 0:
                    print(f"     🔴 {rule}: {count} vezes ({count/len(packets)*100:.1f}%)")
        else:
            print("   ✅ Nenhum alerta (comportamento esperado para tráfego normal)")
        
        print("   " + "-" * 40)

# Executa validação
resultados_validacao = validar_regras()

print("\n" + "=" * 60)
print("✅ Validação concluída!")
````

````python
# Célula 6: Verificador de Sintaxe do Código P4

def verificar_sintaxe_p4():
    """Verifica se o código P4 gerado tem sintaxe válida"""
    
    print("🔍 Verificando sintaxe do código P4 gerado...")
    print("=" * 60)
    
    if not os.path.exists('goose_ids.p4'):
        print("❌ Arquivo goose_ids.p4 não encontrado!")
        return False
    
    with open('goose_ids.p4', 'r') as f:
        codigo = f.read()
    
    linhas = codigo.split('\n')
    erros = []
    
    # Verificações básicas de sintaxe P4
    for i, linha in enumerate(linhas, 1):
        linha_stripped = linha.strip()
        
        # Verifica se há comentários no formato errado
        if '//' in linha and '/*' not in linha and '*/' not in linha:
            # Comentários // são válidos em P4
            pass
        
        # Verifica chaves balanceadas
        if linha_stripped == '{':
            pass
        
        # Verifica se há indentação com tabs (P4 não gosta)
        if '\t' in linha and not linha.startswith(' '):
            erros.append(f"Linha {i}: Usa TAB em vez de espaços")
    
    # Verifica se todas as regras estão presentes
    for regra in regras:
        if regra['name'] not in codigo:
            erros.append(f"Regra '{regra['name']}' não encontrada no código P4")
        elif '_action' not in codigo:
            erros.append(f"Ação para '{regra['name']}' não encontrada")
    
    if erros:
        print("❌ Erros encontrados:")
        for erro in erros:
            print(f"   - {erro}")
        return False
    else:
        print("✅ Sintaxe do código P4 parece válida!")
        print(f"   - {len(regras)} regras implementadas")
        print(f"   - {len([l for l in linhas if 'action' in l and '_action' in l])} ações definidas")
        return True

# Executa verificação
verificar_sintaxe_p4()
````

````python
# Célula 7: Preparação de Scripts para Execução Real

import stat

def criar_scripts_execucao():
    """Cria scripts auxiliares para executar o switch P4"""
    
    # Cria diretório scripts se não existir
    os.makedirs('scripts', exist_ok=True)
    
    # Script de setup
    setup_script = '''#!/bin/bash
# Setup do ambiente P4 para GOOSE IDS

echo "=========================================="
echo "🚀 Configurando ambiente GOOSE IDS P4"
echo "=========================================="

# Verifica dependencias
echo ""
echo "📦 Verificando dependencias..."

if ! command -v p4c &> /dev/null; then
    echo "❌ p4c nao encontrado"
    echo "   Instale com: sudo apt-get install p4lang-p4c"
    exit 1
fi
echo "✅ p4c: $(p4c --version)"

if ! command -v simple_switch &> /dev/null; then
    echo "❌ simple_switch nao encontrado"
    echo "   Instale com: sudo apt-get install p4lang-bmv2"
    exit 1
fi
echo "✅ simple_switch: encontrado"

# Compila o P4
echo ""
echo "🔨 Compilando goose_ids.p4..."

if [ -f "../goose_ids.p4" ]; then
    cd ..
    p4c --target bmv2 --arch v1model goose_ids.p4 -o goose_ids.json
    
    if [ $? -eq 0 ]; then
        echo "✅ Compilacao bem sucedida!"
        echo "📄 Arquivo gerado: goose_ids.json"
    else
        echo "❌ Falha na compilacao"
        exit 1
    fi
    cd scripts
else
    echo "❌ goose_ids.p4 nao encontrado na raiz do projeto"
    exit 1
fi

# Cria interfaces virtuais
echo ""
echo "🔧 Criando interfaces virtuais..."
sudo ip link add veth1 type veth peer name veth2 2>/dev/null
sudo ip link set veth1 up 2>/dev/null
sudo ip link set veth2 up 2>/dev/null

echo ""
echo "=========================================="
echo "✅ Ambiente preparado com sucesso!"
echo "=========================================="
echo ""
echo "Para iniciar o switch:"
echo "  cd .. && sudo simple_switch -i 1@veth1 -i 2@veth2 goose_ids.json"
echo ""
echo "Para testar:"
echo "  sudo tcpreplay -i veth1 traffic_grayhole.pcap"
'''
    
    # Script de teste
    test_script = '''#!/bin/bash
# Script de teste do GOOSE IDS P4

echo "=========================================="
echo "🧪 Testando GOOSE IDS P4"
echo "=========================================="

# Verifica se o arquivo JSON existe
if [ ! -f "../goose_ids.json" ]; then
    echo "❌ goose_ids.json nao encontrado"
    echo "   Execute ./setup_p4.sh primeiro!"
    exit 1
fi

# Verifica se os PCAPs existem
if [ ! -f "../traffic_normal.pcap" ]; then
    echo "❌ Arquivos PCAP nao encontrados"
    echo "   Execute o notebook primeiro para gerar os PCAPs"
    exit 1
fi

# Inicia o switch em background
echo ""
echo "🚀 Iniciando switch P4..."
cd ..
sudo simple_switch -i 1@veth1 -i 2@veth2 goose_ids.json --log-file switch.log &
SWITCH_PID=$!
sleep 3

echo "✅ Switch iniciado (PID: $SWITCH_PID)"
echo ""

# Função para testar tráfego
test_traffic() {
    local name=$1
    local file=$2
    echo "📤 Testando $name..."
    sudo tcpreplay -i veth1 "$file" -q
    sleep 1
}

# Executa testes
test_traffic "traffic normal" "traffic_normal.pcap"
test_traffic "ataque Grayhole" "traffic_grayhole.pcap"
test_traffic "ataque High StNum" "traffic_high_stnum.pcap"
test_traffic "ataque Injection" "traffic_injection.pcap"

# Coleta estatisticas
echo ""
echo "📊 Estatisticas do switch:"
echo "=========================================="

# Tenta conectar ao CLI
echo "table_dump MyIngress.rule_grayhole_table" | simple_switch_CLI --thrift-port 9090 2>/dev/null
echo ""
echo "table_dump MyIngress.rule_high_StNum_state" | simple_switch_CLI --thrift-port 9090 2>/dev/null
echo ""
echo "table_dump MyIngress.rule_injection_seq" | simple_switch_CLI --thrift-port 9090 2>/dev/null

# Para o switch
echo ""
echo "🛑 Parando switch..."
sudo kill $SWITCH_PID 2>/dev/null

echo ""
echo "=========================================="
echo "✅ Teste concluido!"
echo "📄 Logs disponiveis em: switch.log"
echo "=========================================="
cd scripts
'''
    
    # Salva os scripts
    with open('scripts/setup_p4.sh', 'w') as f:
        f.write(setup_script)
    
    with open('scripts/test_p4.sh', 'w') as f:
        f.write(test_script)
    
    # Torna scripts executáveis
    os.chmod('scripts/setup_p4.sh', stat.S_IRWXU)
    os.chmod('scripts/test_p4.sh', stat.S_IRWXU)
    
    print("✅ Scripts de execução criados:")
    print(f"   - scripts/setup_p4.sh (prepara ambiente)")
    print(f"   - scripts/test_p4.sh (executa testes)")
    print("\n📝 Para usar:")
    print(f"   cd scripts")
    print(f"   ./setup_p4.sh")
    print(f"   ./test_p4.sh")

# Cria scripts
criar_scripts_execucao()
````

## Emulação com Mininet:

```python
from mininet.net import Mininet
from mininet.node import Host, Node
from mininet.link import TCLink
from mininet.cli import CLI

class P4Switch(Node):
    def __init__(self, name, json_path, **kwargs):
        Node.__init__(self, name, **kwargs)
        self.json_path = json_path
    
    def start(self):
        self.cmd(f'simple_switch --thrift-port 9090 -i 1@veth1 -i 2@veth2 {self.json_path} &')

# Criar topologia
net = Mininet(link=TCLink)
h1 = net.addHost('h1')
h2 = net.addHost('h2')
s1 = net.addSwitch('s1', cls=P4Switch, json_path='goose_ids.json')
net.addLink(h1, s1)
net.addLink(h2, s1)
net.start()
CLI(net)
net.stop()
```

| simple_switch não encontrado | `sudo apt-get install p4lang-bmv2` |

| Interfaces veth não existem | `sudo modprobe veth` |
